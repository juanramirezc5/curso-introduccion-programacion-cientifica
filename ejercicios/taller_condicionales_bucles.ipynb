## EJERCICIO 6

se recorren los datos para saber si esta agotado, bajo o suficiente

```python
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

for indice in range(len(cantidades_en_stock)):
    cantidad = cantidades_en_stock[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    print(f"Producto {indice + 1}: {nivel}")
```

## EJERCICIO 7

Senrecorre para mirar que dias estan caluorosos por encima de 20

```python
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]
suma_temperaturas = 0
dias_caluorosos = 0

for indice in range(len(temperaturas_semana)):
    temperatura = temperaturas_semana[indice]
    suma_temperaturas += temperatura
    if temperatura > 20:
        dias_caluorosos += 1

promedio_semana = suma_temperaturas / len(temperaturas_semana)

print("Promedio semanal:", promedio_semana)
print("Días caluoroso:", dias_caluorosos)
```

## EJERCICIO 8

se usa un while que se repite mientras el presupuesto sea mayor o igual al precio de un tornillo.

```python
presupuesto_ferreteria = 3000
precio_tornillo = 250
tornillos_comprados = 0

while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria -= precio_tornillo
    tornillos_comprados += 1

print("Tornillos comprados", tornillos_comprados)
print("Presupuesto sin usar:", presupuesto_ferreteria)
```

## EJERCICIO 10

recorre dos listas hasta encontrar una cantidad que deba reponer y ya de ahi finaliza y no sigue con la lista

```python
nombres_herramientas = ["Martillo", "Taladro", "Sierra", "Destornillador", "Nivel", "Llave inglesa"]
cantidades_herramientas = [5, 0, 2, 8, 0, 1]

for indice in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[indice]
    cantidad = cantidades_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue

    if nivel == "Agotado":
        print("ALERTA tienes que hacer reposición urgente de", nombre)
        break

    print(f"Stock bajo: {nombre} ({cantidad} unidades)")
```

## BONUS

En el ejercicio 10 ya no uso break sino que se guardan para mostrarlas al final

```python
herramientas_agotadas = []

for indice in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[indice]
    cantidad = cantidades_herramientas[indice]

    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    if nivel == "Stock suficiente":
        continue

    if nivel == "Agotado":
        herramientas_agotadas.append(nombre)
    else:
        print(f"Stock bajo; {nombre} ({cantidad} unidades)")

print("Herramientas agotadas por reponer:", herramientas_agotadas)
```
