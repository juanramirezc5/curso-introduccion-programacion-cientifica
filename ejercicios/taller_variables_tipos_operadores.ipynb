## **TALLER DE VARIABLES, TIPOS DE DATOS Y OPERADORES**

## EJERCICIO 1


Declarar variables

```python
nombre_producto = "Cuaderno"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print("nombre_producto:", type(nombre_producto))
print("precio_cuaderno:", type(precio_cuaderno))
print("precio_lapiz:", type(precio_lapiz))
print("precio_cafe:", type(precio_cafe))
print("hay_descuento:", type(hay_descuento))
print("cantidad_cuadernos:", type(cantidad_cuadernos))
print("cantidad_lapices:", type(cantidad_lapices))
```

## EJERCICIO 2


calcula el total y ademas involucra porcentaje

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print("Total cuadernos:", total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print("Total lápices:", total_lapices)

total_general = total_cuadernos + total_lapices + precio_cafe
print("Total general:", total_general)

descuento = total_general * 0.10
print("Descuento (10%):", descuento)

total_con_descuento = total_general - descuento
print("Total con descuento:", total_con_descuento)
```

## EJERCICIO 3


se aplica division entera, potencia y modulo

```python
paquetes_completos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
precio_cafe_cuadrado = precio_cafe ** 2

print("Paquetes completos de 4:", paquetes_completos)
print("Cuadernos sueltos:", cuadernos_sueltos)
print("Precio del café al cuadrado:", precio_cafe_cuadrado)
```

## EJERCICIO 5


se aplica el promedio y realoiza una comparacion

```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

temp_promedio = (temp_lunes + temp_martes + temp_miercoles) / 3
print("Promedio de los tres días:", temp_promedio)

diferencia_martes_lunes = temp_martes - temp_lunes
print("Diferencia martes - lunes:", diferencia_martes_lunes)

print("¿Martes más caluroso que el promedio de referencia?", temp_martes > temp_promedio_semana)
print("¿Miércoles igual al promedio de referencia?", temp_miercoles == temp_promedio_semana)

lunes_mas_frio = temp_lunes < temp_martes and temp_lunes < temp_miercoles
print("¿Lunes fue el día más frío?", lunes_mas_frio)
```
