## Ejercicio 1

Se crea una función que corrige una lectura con la fórmula (valor * pendiente + intercepto). Si no se le dan la pendiente y el intercepto, deja la lectura igual

```python
def calibrar_lectura(valor, pendiente=1.0, intercepto=0.0):
    return valor * pendiente + intercepto

sin_calibrar = calibrar_lectura(10.0)
calibrada = calibrar_lectura(10.0, 1.2, -0.5)
con_intercepto = calibrar_lectura(10.0, intercepto=2.0)

print(sin_calibrar)
print(calibrada)
print(con_intercepto)
```

## Ejercicio 2

Se crea una función que dice si una lectura sirve o no. Sirve si no es -999.0 y está entre 0 y 500

```python
lecturas_sbc_a = [10.0, -999.0, 25.0, 40.0, -999.0, 8.0, 52.0, 600.0]

def es_lectura_valida(valor):
    return valor != -999.0 and 0 <= valor <= 500.0

print(es_lectura_valida(25.0))
print(es_lectura_valida(-999.0))
print(es_lectura_valida(600.0))
print(es_lectura_valida(0.0))
```

## Ejercicio 3

se crea una función que recorre la lista y suma solo las lecturas que sirven, luego, calcula el promedio si ninguna sirve, devuelve None.

```python
def promedio_valido(lecturas):
    suma = 0.0
    cantidad = 0
    for lectura in lecturas:
        if es_lectura_valida(lectura):
            suma += lectura
            cantidad += 1
    if cantidad == 0:
        return None
    return suma / cantidad

promedio_sbc_a = promedio_valido(lecturas_sbc_a)

print(promedio_sbc_a)
print(promedio_valido([-999.0, -999.0]))
```

## Ejercicio 4

Primero se dejan solo las lecturas que sirven después se calibran con map y lambda. Al final secuentan las que pasan de 35.4. No se usan ciclos for.

```python
pendiente_sbc_a = 1.2
intercepto_sbc_a = -0.5

lecturas_utiles = filter(es_lectura_valida, lecturas_sbc_a)
lecturas_calibradas = list(map(lambda v: calibrar_lectura(v, pendiente_sbc_a, intercepto_sbc_a), lecturas_utiles))
excedencias = len(list(filter(lambda v: v > 35.4, lecturas_calibradas)))

print(lecturas_calibradas)
print(excedencias)
```

## Ejercicio 5

se crea una clase que guarda el valor, la unidad y el código del sensor tiene un método para pasar de mg/m3 a ug/m3 y otro para saber si el dato es -999.0



```python
class Lectura:

    def __init__(self, valor, unidad="ug/m3", codigo_sensor="sin asignar"):
        self.valor = valor
        self.unidad = unidad
        self.codigo_sensor = codigo_sensor

    def a_microgramos(self):
        if self.unidad == "mg/m3":
            return self.valor * 1000
        return self.valor

    def es_dato_faltante(self):
        return self.valor == -999.0

lectura_campo = Lectura(0.0252, "mg/m3", "SBC-A")
lectura_faltante = Lectura(-999.0, codigo_sensor="SBC-A")
lectura_normal = Lectura(30.0)

print(lectura_campo.a_microgramos())
print(lectura_faltante.es_dato_faltante())
print(lectura_campo.es_dato_faltante())
print(lectura_normal.unidad)
print(lectura_normal.codigo_sensor)
```

## Ejercicio 6

se crea una clase que guarda el codigo, la pendiente y el intercepto del sensor si el codigo está vacío o la pendiente es cero o negativa, da un error tambien tiene un método para calibrar y dos para mostrarse como textp

```python
class Sensor:

    def __init__(self, codigo, pendiente=1.0, intercepto=0.0):
        if codigo == "":
            raise ValueError("El codigo del sensor no puede estar vacio")
        if pendiente <= 0:
            raise ValueError("La pendiente debe ser mayor que cero")
        self.codigo = codigo
        self._pendiente = pendiente
        self._intercepto = intercepto

    def calibrar(self, valor):
        return calibrar_lectura(valor, self._pendiente, self._intercepto)

    def __repr__(self):
        return f"Sensor(codigo='{self.codigo}', pendiente={self._pendiente}, intercepto={self._intercepto})"

    def __str__(self):
        return f"Sensor {self.codigo}: pendiente {self._pendiente}, intercepto {self._intercepto}"

sensor_a = Sensor("SBC-A", 1.2, -0.5)
sensor_b = Sensor("SBC-B")

print(sensor_a)
print([sensor_a, sensor_b])
print(sensor_a.calibrar(10.0))
print(sensor_b.calibrar(10.0))
```

## Ejercicio 7

se crea una clase que guarda el codigo, el sector y una lista de lecturas. Tiene un metodo para agregar lecturas y otro para sacart todos los valores en ug/m3 luego se llena la estación EST-04 con un for

```python
class Estacion:

    def __init__(self, codigo, sector):
        self.codigo = codigo
        self.sector = sector
        self._lecturas = []

    def agregar(self, lectura):
        self._lecturas.append(lectura)

    def valores(self):
        lista_valores = []
        for lectura in self._lecturas:
            lista_valores.append(lectura.a_microgramos())
        return lista_valores

    def __repr__(self):
        return f"Estacion('{self.codigo}', '{self.sector}', lecturas={len(self._lecturas)})"

datos_campana = [
    (10.0, "ug/m3"), (-999.0, "ug/m3"), (0.025, "mg/m3"), (40.0, "ug/m3"),
    (-999.0, "ug/m3"), (8.0, "ug/m3"), (52.0, "ug/m3"), (600.0, "ug/m3"),
]

estacion_oriente = Estacion("EST-04", "Oriente")

for valor, unidad in datos_campana:
    estacion_oriente.agregar(Lectura(valor, unidad, "SBC-A"))

print(estacion_oriente)
print(estacion_oriente.valores())
```

## Ejercicio 8

Se crea una función que dice la categoría del aire según la concentración. También se crea una clase que calcula el promedio calibrado de la estación y muestra el resultado con su categoría.

```python
def categoria_pm25(concentracion):
    if concentracion <= 12.0:
        return "Buena"
    elif concentracion <= 35.4:
        return "Moderada"
    elif concentracion <= 55.4:
        return "Dañina para grupos sensibles"
    else:
        return "Dañina para la salud"

class InformeEstacion:

    def __init__(self, estacion, sensor):
        self._estacion = estacion
        self._sensor = sensor

    def promedio_calibrado(self):
        validas = filter(es_lectura_valida, self._estacion.valores())
        calibradas = list(map(self._sensor.calibrar, validas))
        return promedio_valido(calibradas)

    def categoria(self):
        return categoria_pm25(self.promedio_calibrado())

    def __str__(self):
        promedio = round(self.promedio_calibrado(), 1)
        return f"{self._estacion.codigo}: {promedio} ug/m3 - {self.categoria()}"

informe = InformeEstacion(estacion_oriente, sensor_a)

print(informe)
print(informe.promedio_calibrado())
```
