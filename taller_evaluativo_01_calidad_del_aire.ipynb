```python
codigo_estacion = "EST-01"
sector = "Norte"
altitud_m = 1495
numero_lecturas = 305
latitud = 6.28
sensor_calibrado = True
factor_conversion = 1000

print(type(codigo_estacion))
print(type(sector))
print(type(altitud_m))
print(type(numero_lecturas))
print(type(latitud))
print(type(sensor_calibrado))
print(type(factor_conversion))
```

### EJERCICIO 2

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print(lectura_sensor_ug)
print(error_absoluto)
print(error_relativo_porcentual)
print(error_cuadratico)
```

### EJERCICIO 4

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor
lectura_valida = esta_en_rango and not es_dato_faltante

print(es_dato_faltante)
print(esta_en_rango)
print(lectura_valida)
```

### EJERCICIO 6

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura_actual in lecturas_pm25:
    if lectura_actual == -999.0:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(lectura_actual)
    suma_pm25 += lectura_actual

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print(len(lecturas_validas))
print(lecturas_descartadas)
print(promedio_pm25)
```

### EJERCICIO 7

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for lectura_actual in lecturas_validas:
    if lectura_actual > maximo_pm25:
        maximo_pm25 = lectura_actual
    if lectura_actual < minimo_pm25:
        minimo_pm25 = lectura_actual

rango = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0
for lectura_actual in lecturas_validas:
    suma_desviaciones += (lectura_actual - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print(maximo_pm25)
print(minimo_pm25)
print(rango)
print(desviacion_estandar)
```

### EJERCICIO 8

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0
horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print(horas_transcurridas)
print(concentracion)
```

### EJERCICIO 9

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud, longitud = estacion["coordenadas"]

    suma_lecturas_estacion = 0.0
    for valor_lectura in estacion["lecturas"]:
        suma_lecturas_estacion += valor_lectura
    promedio_estacion = suma_lecturas_estacion / len(estacion["lecturas"])

    humedad = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(estacion["codigo"], estacion["sector"], latitud, longitud, promedio_estacion, categoria_estacion, humedad)
```
